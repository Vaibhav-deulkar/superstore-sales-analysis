```python
df = df.drop_duplicates()
```

##Correlation (Numerical Columns)


```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.show()
```

##Sales by Category


```python
plt.figure(figsize=(8,5))
sns.barplot(x="Category", y="Sales", data=df)
plt.title("Sales by Category")
plt.show()
```

##Profit by Category

```python
plt.Figure(figsize=(8,5))
sns.barplot(x="Category", y="Profit", data=df)
plt.title("Profit by Category")
plt.show()
```

##Sales by Region

```python
plt.Figure(figsize=(8,5))
sns.barplot(x="Region", y="Sales",data=df)
plt.title("Sales by Region")
plt.show()
```

##Sales by Segment

```python
sns.barplot(x="Segment", y="Sales", data=df)
plt.show()
```

Discount vs Profit

```python
sns.scatterplot(x="Discount", y="Profit", data=df)
plt.show()
```

## Product

```python
top_products = df.groupby("Sub-Category")["Sales"].sum().sort_values(ascending=False).head(10)

top_products.plot(kind="bar")
plt.title("Top 10 Sub-Categories by Sales")
```

##State-wise Sales


```python
state_sales = df.groupby("State")["Sales"].sum().sort_values(ascending=False).head(10)

state_sales.plot(kind="bar")
```

## Business Insights

- Technology generated the highest sales.
- Office Supplies had the lowest profit margin.
- High discounts often resulted in negative profits.
- The West region achieved the highest revenue.
- Consumer segment contributed the most sales.

## Conclusion

This analysis helped identify profitable categories, loss-making products, regional performance, and the impact of discounts on profitability. The findings can support better pricing and inventory decisions.
